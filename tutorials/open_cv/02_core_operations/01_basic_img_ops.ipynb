```python
b, g, r = cv.split(img)
for channel in [b, g, r]:
    plt.imshow(channel, cmap='gray')
    plt.show();
```

```python
blue = cv.merge((b, 0*b, 0*b))
green = cv.merge((0*g, g, 0*g))
red = cv.merge((0*r, 0*r, r))
for i in (blue, green, red):
    plt.imshow(i)
    plt.show()
```

```python
rot = cv.merge((g, r, b))
plt.imshow(rot);
```

### Making borders (padding)

```python
plt.figure(figsize=(10, 10))
N = 50
BLUE = (0, 0, 255)
replicate = cv.copyMakeBorder(img, N, N, N, N, cv.BORDER_REPLICATE)
reflect = cv.copyMakeBorder(img, N, N, N, N, cv.BORDER_REFLECT)
reflect101 = cv.copyMakeBorder(img, N, N, N, N, cv.BORDER_REFLECT_101)
wrap = cv.copyMakeBorder(img, N, N, N, N, cv.BORDER_WRAP)
constant= cv.copyMakeBorder(img, N, N, N, N, cv.BORDER_CONSTANT, value=BLUE)
 
plt.subplot(321)
plt.imshow(img, 'gray')
plt.title('ORIGINAL')

plt.subplot(322)
plt.imshow(replicate, 'gray')
plt.title('REPLICATE')

plt.subplot(323)
plt.imshow(reflect, 'gray')
plt.title('REFLECT')

plt.subplot(324)
plt.imshow(reflect101, 'gray')
plt.title('REFLECT_101')

plt.subplot(325)
plt.imshow(wrap, 'gray')
plt.title('WRAP')

plt.subplot(326)
plt.imshow(constant, 'gray')
plt.title('CONSTANT');
plt.tight_layout()
```

```python
rand = np.int32(np.random.uniform(1, 255, (4, 4, 3)))
plt.imshow(rand);
```

```python
N  = 2
reflect = cv.copyMakeBorder(rand, N, N, N, N, cv.BORDER_REFLECT)
reflect101 = cv.copyMakeBorder(rand, N, N, N, N, cv.BORDER_REFLECT_101)

plt.subplot(121)
plt.imshow(reflect, 'gray')
plt.title('REFLECT')

plt.subplot(122)
plt.imshow(reflect101, 'gray')
plt.title('REFLECT_101')

plt.tight_layout();
```
