# Draw a Line, Rectangle, Circle, Ellipse, Polygon, or Text

```python
img = np.zeros((512, 512, 3), np.uint8)
cv.line(
    img, 
    (0, 0),       # start pt
    (511, 511),   # end pt
    (255, 0, 0),  # color 
    5             # line width
)

cv.rectangle(
    img, 
    (384, 4),     # top left
    (510, 128),   # bottom right
    (0, 255, 0),  # color
    3)

cv.circle(
    img,
    (447, 67),  # center
    63,         # radius
    (0, 0, 255),
    -1)         # fill

cv.ellipse(
    img,
    (256, 256),  # center
    (100, 50),   # maj axis, min axis lens
    45,          # angle of rot
    90,          # start of curve in angle
    270,         # end of curve
    (255, 255, 0),
    4)

pts = np.array([[10, 5], [20, 30], [70, 20], [50,10]], np.int32)
pt = pts.reshape((-1, 1, 2))
cv.polylines(
    img, 
    [pts], 
    True,  # close?
    (0, 255, 255),
    3)

font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(
    img, 
    'Amaze!', 
    (10, 500),   # bottom left of text
    font, 
    4,           # font scale
    (255, 0, 255),
    3,
    cv.LINE_AA)  # anti-alias
plt.imshow(img);
```
