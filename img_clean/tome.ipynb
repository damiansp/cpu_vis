```python
IMG = '../img/bevpics'
```

```python
def equalize_hist(img):
    grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    eq = cv2.equalizeHist(grey)
    return eq
```

```python
img = cv2.imread(f'{IMG}/rphs.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img);
```

```python
img = img[700:3400, 500:2300, :]
plt.imshow(img);
```

```python
eq = equalize_hist(img)
plt.imshow(eq, 'grey');
```

```python
def stretch_contrast(img):
    img_min = np.min(img)
    img_max = np.max(img)
    return (img - img_min) * (255 // (img_max - img_min))
```

```python
stretched = stretch_contrast(img)
plt.imshow(stretched);
```
